```python
import pandas as pd
import glob
import random

INPUT_FEATURES = ['FireMask','elevation', 'th', 'vs',  'tmmn', 'tmmx', 'sph', 
                    'pr', 'pdsi', 'NDVI', 'population', 'erc', 'PrevFireMask']

def combineFiles(quantity, infolder, outfolder, inputFeatures):

    # List of CSV files to combine (you can adjust the path/pattern)
    all_files = glob.glob(f"{infolder}/*.csv")  # e.g., all CSVs in 'data/' folder
    max = len(all_files)

    if quantity > max:
        print(f"Too many files requested, max {max}")
        return

    rand_nums = random.sample(range(19), quantity)
    rand_files = [f"{infolder}/{n:02d}.csv" for n in rand_nums]

    combined_df = pd.DataFrame()
    for f in rand_files:
        curr = pd.read_csv(f)
        curr = curr[inputFeatures]
        combined_df = pd.concat([combined_df, curr])

    # Save to a single CSV
    combined_df.to_csv(f"{outfolder}/split{quantity:d}.csv", index=False)

    print(f"Combined {quantity} files into 'split{quantity}.csv'.")


combineFiles(19, "renamed-raw-csvs", "combo-files-for-norm", INPUT_FEATURES)
```
